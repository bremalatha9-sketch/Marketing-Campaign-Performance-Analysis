```python
# Handle missing values ( drop)
# Remove duplicate
df = df.drop_duplicates()

# Check duplicates again
print("Duplicates:", df.duplicated().sum())
df.shape
```

```python
# Remove duplicate in Compaign id
df[df['Campaign_ID'].duplicated()]              #keeping the first occurence
df = df.drop_duplicates(subset=['Campaign_ID'])
df['Campaign_ID'] = range(1, len(df) + 1)        #generated new sequential cam id
df.shape
```

```python
# Rename column names
df.columns = df.columns.str.strip().str.lower() #it make column name simple and consistent

# Check updated column names
print(df.columns)
```

```python
print(df["duration"])
```

```python
# Convert duration into numeric days
df['duration'] = pd.to_numeric(
    df['duration'].astype(str).str.extract(r'(\d+)')[0],# extracted numeric values further numerical calculations and insights.
    errors='coerce'
)
```

```python
#Renamed duration to duration_days for understanding the columns given in days
df.rename(columns={'duration': 'duration_days'}, inplace=True)
```

```python
#convert convertion_rate into numeric
df["conversion_rate"] = pd.to_numeric(df["conversion_rate"], errors="coerce")
df["conversion_rate"]
```

```python
df['conversion_rate'] = df['conversion_rate'] * 100
df["conversion_rate"]
```

```python
# convert date into datetime
df["date"]=pd.to_datetime(df["date"])
```

```python
print(df["acquisition_cost"])
```

```python

# Remove dollar sign and comma for further numerical calculations and insights

df['acquisition_cost'] = pd.to_numeric(
    df['acquisition_cost'].astype(str)
    .str.replace('$', '', regex=False)  #regex it treats given value in normal text
    .str.replace(',', '', regex=False),
    errors='coerce'                     #converts invalid values into nan
)
```

```python
df.rename(columns={"acquisition_cost":"acquisition_cost($)"},inplace=True)
```

```python
print(df.dtypes)
```

```python
# Numerical columns filled missing values
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Corrected inconsistent text formatting in categorical columns
# Removed unwanted spaces and converted text into proper title case.
text_columns = [
    'company',
    'campaign_type',
    'target_audience',
    'channel_used',
    'location',
    'language',
    'customer_segment'
]

for col in text_columns:
    df[col] = df[col].astype('string').str.strip().str.title()
```

```python
# To identify different categories or values present in each categorical column(unique values).
cat_cols = [
    'company', 'campaign_type', 'language',
    'channel_used', 'customer_segment',
    'location', 'target_audience'
]

for col in cat_cols:
    print(col, ":", df[col].unique())
```

```python
# Fill missing values in categorical columns using mode
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python

 #Fill missing dates with the most frequent date
df["date"]=df["date"].fillna(df["date"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
# Feature transformatins for further calculations and insights(trends and patterns)
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['quarter'] = df['date'].dt.quarter
df['day'] = df['date'].dt.day
df['day_name'] = df['date'].dt.day_name()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
# Recalculate CTR (Click-Through Rate) using clicks and impressions
df['ctr%'] = ((df['clicks'] / df['impressions']) * 100).round(2)


print(df["ctr%"])
```

```python

# CPC (Cost Per Click)
df['cpc'] = (df['acquisition_cost($)'] / df['clicks'].replace(0, np.nan)).round(2)

df[ 'cpc'].head()
```

```python
# Estimate campaign revenue using acquisition cost and ROI
df['revenue($)'] = df["acquisition_cost($)"] * (1 + df['roi']).round(2) # To compare a campaign performance
```

```python
df["profit($)"] = (df["revenue($)"] - df["acquisition_cost($)"]).round(2)
```

```python
# Create a new column called 'cost_band' to categorize campaigns based on acquisition cost
df["cost_band"] = pd.qcut(
    df["acquisition_cost($)"],
    4,                               # Divide the data into 4 equal-sized groups (quartiles)
    labels=[                         # Assign labels to each quartile
        "Q1 (lowest)",                # 25% campaigns with the lowest acquisition cost
        "Q2",                         # Next 25% campaigns
        "Q3",
        "Q4 (highest)"                # 25% campaigns with the highest acquisition cost
    ]
)

df["cost_band"]
```

```python
# Create a new column called 'roi_band' to categorize campaigns based on their ROI
df["roi_band"] = pd.cut(
    df["roi"],                      # Select the ROI column
    [0, 4, 6, 8],                  # Define the ROI range boundaries
    labels=[                       # Assign category names to each ROI range
        "Low ROI (2-4)",
        "Mid ROI (4-6)",
        "High ROI (6-8)"
    ]
)
df["roi_band"]
```

```python
# Check CTR range
print("Minimum CTR:", df['ctr%'].min())
print("Maximum CTR:", df['ctr%'].max())

# Create CTR bands
df['CTR_band'] = pd.cut(
    df['ctr%'],
    bins=[-float('inf'), 1, 3, 5, float('inf')],
    labels=['Low CTR', 'Medium CTR', 'High CTR', 'Very High CTR']
)

# Check campaign distribution
print(df['CTR_band'].value_counts())
```

```python
print("Total Campaign Records:", len(df))

print("Average Conversion Rate:",
      round(df['conversion_rate'].mean(),2),
      "%")

print("Average ROI:",
      round(df['roi'].mean(),2))

print("Total Clicks:",
      df['clicks'].sum())

print("Total Impressions:",
      df['impressions'].sum())
```

```python
# Select numerical columns

num_cols = df.select_dtypes(include=np.number).columns
# Calculate the IQR(interquartile range) for each numerical column
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) | (df[col] > upper)# identify high,low value in the dataset
    ]


    print(col, "Outliers:", len(outliers))
```

```python
# Select columns(Outlier treatment)
cols = ['ctr%', 'cpc', 'revenue($)',"profit($)"]

for col in cols:

    # Calculate Q1 and Q3
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Find lower and upper limits
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Count outliers before capping
    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, ":", len(outliers), "outliers")

    # Cap outliers
    df[col] = df[col].clip(lower, upper)
```

```python
# Check skewness to understand the distribution of numerical columns
# To check whether the numerical data is balanced or skewed to one side.
num_cols = df.select_dtypes(include=np.number).columns
print("Skewness:")
print(df[num_cols].skew())
```

```python
#importing essential libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/Marketing_Campaign_Performance_Cleaned.csv")
```

```python
# Stage 3 – EDA and Visualizations
# Univariate analysis: Understand one column at a time.
#  1.Campaign type distribution – Countplot


plt.figure(figsize=(8, 6))  # Set the chart size

# Create a countplot to show the number of campaigns for each campaign type
ax = sns.countplot(
    data=df,
    x='campaign_type',
    hue="campaign_type",
    order=df['campaign_type'].value_counts().index,  # Arrange campaign types by highest count
    legend=False,  # Hide the legend
    palette='Set2'  # Apply different colours
)

# Add total count labels on top of each bar
for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=8, fontweight='bold')

# Add chart title and axis labels
plt.title("Campaign Type Distribution")
plt.xlabel("Campaign Type")
plt.ylabel("Number of Campaigns")

# Rotate campaign names for better readability
plt.xticks(rotation=45)
```

```python
# Chart 3 – Revenue Distribution (Histogram)

fig,ax=plt.subplots(figsize=(8, 5))  # Set the chart size

# Create a histogram to show the distribution of campaign revenue
sns.histplot(data=df, x="revenue($)", bins=40, kde=True,ax=ax) #Divides revenue values into 30 groups.
                                                         #Adds a smooth distribution curve.
# Add chart title and axis labels
ax.axvline(df["revenue($)"].median(), color="red", ls="--", label=f"Median ${df['revenue($)'].median():,.0f}")
ax.axvline(df["revenue($)"].mean(), color="orange", ls="--", label=f"Mean ${df['revenue($)'].mean():,.0f}")
ax.set_title("Distribution of Campaign Revenue")
ax.set_xlabel("Revenue ($)")
ax.set_ylabel("Frequency")

# Adjust chart spacing
plt.tight_layout()

# Display the chart
plt.show()
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
# Calculate monthly average revenue
monthly = df.groupby(
    df['date'].dt.to_period('M')
)['revenue($)'].mean()

monthly.index = monthly.index.to_timestamp()

# Create chart
plt.figure(figsize=(11, 5))

# Plot revenue trend
plt.plot(
    monthly.index,
    monthly.values,
    color='#3a9d5d',
    marker='o',
    linewidth=2.5
)

# Fill area below the curve
plt.fill_between(
    monthly.index,
    monthly.values,
    monthly.min() * 0.95,
    color='#3a9d5d',
    alpha=0.15
)

# Add yearly average line
plt.axhline(
    df['revenue($)'].mean(),
    color='red',
    linestyle='--',
    label=f"Yearly Average ${df['revenue($)'].mean():,.0f}"
)

# Add title and labels
plt.title('Monthly Average Revenue Trend (2021)', fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Average Revenue per Campaign ($)')

# Display month names
plt.xticks(
    monthly.index,
    monthly.index.strftime('%b')
)

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
cols = {
    'roi': ('ROI', '#2a7fba'),
    'ctr%': ('CTR (%)', '#3a9d5d'),
    'cpc': ('CPC ($)', '#e07b39')
}

sns.set_theme(style='whitegrid')

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, (col, (label, color)) in zip(axes, cols.items()):

    # Remove top 1% extreme values for visualization
    data = df[col].dropna()
    upper = data.quantile(0.99)
    plot_data = data[data <= upper]

    # Create histogram
    sns.histplot(
        plot_data,
        bins=30,
        kde=True,
        color=color,
        edgecolor='white',
        ax=ax
    )

    # Mean line
    ax.axvline(
        plot_data.mean(),
        color='red',
        ls='--',
        label=f'Mean {plot_data.mean():.2f}'
    )

    # Median line
    ax.axvline(
        plot_data.median(),
        color='black',
        ls=':',
        label=f'Median {plot_data.median():.2f}'
    )

    ax.set_title(f'Distribution of {label}',
                 color=color, fontweight='bold')

    ax.set_xlabel(label)
    ax.set_ylabel('Count')
    ax.legend()

plt.suptitle('Distribution of ROI, CTR and CPC',
             fontsize=15, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Convert date column into datetime format
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Extract month from date
df['month'] = df['date'].dt.to_period('M').dt.to_timestamp()

# Calculate monthly clicks and average conversion rate
monthly = df.groupby('month').agg({
    'clicks': 'sum',
    'conversion_rate': 'mean'
}).reset_index()

# Calculate estimated conversions
monthly['conversions'] = (
    monthly['clicks'] * monthly['conversion_rate'] / 100
)

# Set chart style
sns.set_theme(style='whitegrid')

fig, ax1 = plt.subplots(figsize=(12, 6))

# Bar chart for total clicks
ax1.bar(
    monthly['month'].dt.strftime('%b'),
    monthly['clicks'],
    color='purple',
    label='Total Clicks'
)

ax1.set_xlabel('Month')
ax1.set_ylabel('Total Clicks', color='purple')

# Line chart for estimated conversions
ax2 = ax1.twinx()

ax2.plot(
    monthly['month'].dt.strftime('%b'),
    monthly['conversions'],
    color='lightblue',
    marker='D',
    linewidth=2,
    markersize=6,
    label='Estimated Conversions'
)

ax2.set_ylabel('Estimated Conversions', color='blue')

# Add title
plt.title(
    'Monthly Total Clicks vs Estimated Conversions',
    fontweight='bold'
)

# Combine legends
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.tight_layout()
plt.show()
```
